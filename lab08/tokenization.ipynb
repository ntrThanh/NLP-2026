# Lab 08 — Calculation Exercise 2: Tokenization

Môn học: Xử lý ngôn ngữ tự nhiên và ứng dụng  
Bài tập tính toán 2: Phân tích Tokenization với mô hình BERT ()


```
from transformers import AutoTokenizer

model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
text = "The cats are playing."
print(f"Loaded tokenizer: {model_name}")
print(f"Input text: '{text}'")
```

## 1 & 2. Tokenize và Token Sequence

Tokenize câu đầu vào bằng tokenizer được cung cấp.


```
# 1. Tokenize tách từ/subwords (chưa có special tokens)
raw_tokens = tokenizer.tokenize(text)
print("1. Raw tokens (subwords):", raw_tokens)

# 2. Tokenize hoàn chỉnh (có special tokens [CLS] và [SEP])
encoded = tokenizer(text)
full_tokens = tokenizer.convert_ids_to_tokens(encoded["input_ids"])
print("2. Full token sequence đưa vào BERT:", full_tokens)
```

## 3, 4 & 5. Token IDs, Special Tokens và Attention Mask


```
input_ids = encoded["input_ids"]
attention_mask = encoded["attention_mask"]

print("3. Token IDs (input_ids):", input_ids)
print("5. Attention mask (không padding):", attention_mask)

print("
4. Xác định Special Tokens:")
for idx, (token, tid) in enumerate(zip(full_tokens, input_ids)):
    if token in ["[CLS]", "[SEP]", "[PAD]", "[MASK]", "[UNK]"]:
        print(f"   - Special token '{token}' tại vị trí {idx} với ID = {tid}")

# Minh họa trường hợp có padding
padded = tokenizer(text, padding="max_length", max_length=10)
print("
--- Minh họa khi có Padding (max_length = 10) ---")
print("Padded tokens:", tokenizer.convert_ids_to_tokens(padded["input_ids"]))
print("Padded input_ids:", padded["input_ids"])
print("Padded attention_mask:", padded["attention_mask"])
```

## 6. So sánh số lượng: words, tokens, subword tokens

- **Words (Từ ngữ nghĩa):** 4 từ ("The", "cats", "are", "playing") và 1 dấu câu ".".
- **Tokens:** 7 tokens nếu tính cả special tokens ([CLS], the, cats, are, playing, ., [SEP]) hoặc 5 tokens nội dung.
- **Subword tokens:** 5 đơn vị subword. Các từ này đều rất phổ biến nên đã có sẵn nguyên từ trong bộ từ vựng 30.522 từ của BERT (không bị bẻ vụn mang tiền tố ##).


```
words = [w.strip(".") for w in text.split()]
print(f"Số words (từ tách theo khoảng trắng): {len(words)} -> {words}")
print(f"Số tokens (gồm cả special tokens): {len(full_tokens)} -> {full_tokens}")
print(f"Số subwords (nội dung): {len(raw_tokens)} -> {raw_tokens}")
```
