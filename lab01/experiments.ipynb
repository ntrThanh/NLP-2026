# 7.1

```python
import pandas as pd 

df = pd.read_json('/media/trong-thanh/Data/University/Fourth Year/NLP/experiments/ex1/data.json', lines = True)
df.head()
```

# 7.2

```python
df['clean_text'] = (
    df['text']
    .fillna("")
    .str.lower()
    .str.replace(r'[^a-z0-9\s]', ' ', regex=True)
    .str.replace(r'\s+', ' ', regex=True)
    .str.strip()
)
df.head()
```

# 7.3

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
X_counter = vectorizer.fit_transform(df['clean_text'])
print(f'Shape: {X_counter.shape}')
print(X_counter)
```

# 7.4

```python
N, V = X_counter.shape
print(f'N = {N}, V = {V}')
```

```python
nnz = X_counter.nnz
total_elements = N * V
S = 1 - (nnz / total_elements)

print(f"Số lượng văn bản (N): {N:,}")
print(f"Kích thước từ điển (V): {V:,}")
print(f"Tổng số phần tử trong ma trận (N * V): {total_elements:,}")
print(f"Số phần tử khác 0 (nnz): {nnz:,}")
print(f"Tỷ lệ phần tử khác 0: {(nnz / total_elements) * 100:.4f}%")
print(f"Độ thưa Sparsity (S): {S:.6f} (tương đương {S * 100:.4f}%)")
```

# 7.5

```python
import numpy as np

feature_names = np.array(vectorizer.get_feature_names_out())
df_counts = np.array((X_counter > 0).sum(axis = 0)).flatten()

top_20 = np.argsort(df_counts)[::-1][:20]
print('Top 20 terns')
for rank, index in enumerate(top_20, 1):
    print(f'{rank:2d}. {feature_names[index]:<12}')
```

```python
from sklearn.feature_extraction.text import TfidfTransformer

tfidf_transformer = TfidfTransformer()
X_tfidf = tfidf_transformer.fit_transform(X_counter)
idfs = tfidf_transformer.idf_
```

```python
top_20_idf = np.argsort(idfs)[::-1][:20]
print('Top 20 IDF')
for rank, index in enumerate(top_20_idf, 1):
    print(f'{rank:2d}. {feature_names[index]:<12}')
```

```python
tfidf_0 = X_tfidf.getrow(0).toarray().flatten()
top_20_tfidf_0 = np.argsort(tfidf_0)[::-1][:20]

for rank, index in enumerate(top_20_tfidf_0, 1):
    if tfidf_0[index] > 0:
        print(f'{rank:2d}. {feature_names[index]:<12}')
```

Trả lời:

- Từ xuất hienj nhiều trong toàn bộ corpus, khi đó DF = N, khi đó tỉ lệ N / DF = 1 => log (N / DF) = 0, khi đó xuất hiện nhiều lần trong văn bản, trọng số vẫn kéo gần về bằng 0.

- Một từ IDF cao => nó hiếm trong corpus. Đối với từ hiện đại trong văn bản, số lần xuất hiện bằng 0 => TF = 0 khi đó TF IDF = 0, từ đó IDF cao chỉ có giá trị cao trong chính tài liệu hiếm chứa nó.
